# Estimación puntual

En los datasets usados para el laboratorio, sólo trabajamos con muestras de una población que puede ser muy grande, como es el caso de este grupo, que se trabaja con toda la población de la humanidad. Por lo que se introduce el concepto de estimadores, estos son, un valor numérico calculado a partir de la muestra, que permite aproximar un parámetro desconocido de la población. En este caso, la razón de usar estimadores, es por que se hace imposible encuestar a toda la población de la humanidad

La estimación puntual, posee propiedades que sirven para conocer la confiabilidad de los estimadores, estos son el sesgo, la consistencia y la eficiencia.

El sesgo se encarga de verificar si el estimador tiene algún error sistemático, es decir, si aproxima hacia arriba o hacia abajo, este se calcula con la siguiente fórmula

$$\text{Sesgo}(\hat{\theta}) = E[\hat{\theta}] - \theta$$

Con

- $\hat{\theta}$: El estimador usado.
- ${\theta}$: El valor real de la población.

La consistencia mide la cercanía del estimador al valor real si tuviesemos una muestra cada vez más grande, es decir

$$\hat{\theta}_n \xrightarrow{p} \theta \quad \text{cuando } n \to \infty$$

Dado que existen muchos estimadores puntuales, la eficiencia permite comparar estimadores y seleccionar aquel que presenta menor error, entre los estimadores, selecciona y compara, cuál es el que menos se equivoca, para comparar estimadores, podemos calcular su certeza con el error cuadrático medio, donde a menor es el error cuadrático medio, mas eficiente es el estimador

$$\text{ECM}(\hat{\theta}) = \text{Var}(\hat{\theta}) + [\text{Sesgo}(\hat{\theta})]^2$$

## Estimadores a usar

En este laboratorio se usarán dos estimadores puntuales, la media muestral y la proporción muestral.

La media muestral, se utiliza para estimar la media poblacional $𝞵$ y se define como:
$$\bar{X} = \frac{1}{n}\sum_{i=1}^n x_i$$

Con:

- $\bar{X}$: La media muestral.
- $n$: Tamaño de la muestra.
- $x_i$: Observaciones de la muestra.

Sesgo para la media muestral:

${Sesgo}(\bar{X}) = E[\bar{X}] - \mu = \mu - \mu = 0$

Por lo que la media muestral es un estimador insesgado.

Consistencia para la media muestral:
Para la consistencia de la media, debemos medir si la varianza de la media, tiende a cero, cuando el tamaño de la muestra tiende al infinito.

$Var(\bar{X}) = \frac{\sigma^2}{n}$, luego $Var(\bar{X}) = 0$ si $n → ∞$. Con $\sigma^2$ la varianza poblacional.

Eficiencia para la media muestral:

$ECM(\bar{X}) = Var(\bar{X}) + {Sesgo}(\bar{X})^2 = Var(\bar{X}) = \frac{\sigma^2}{n}$

La media muestral es relevante para la pregunta planteada, ya que permite estimar el nivel promedio de estrés para distintos subconjuntos de la muestra, por ejemplo, segmentando según rangos de horas de pantalla, calidad del sueño u ocupación. Esto permite identificar tendencias generales y comparar cómo varía el estrés en función de estas variables, lo cual es fundamental para analizar la relación planteada en la pregunta de investigación.

La proporción muestral, se utiliza para estimar la proporción poblacional $p$, y se define como:
$$\hat{p} = \frac{\text{X}}{n}$$
Con:
- $X$: Número de casos que cumplen la condición.
- $n$: Tamaño de la muestra.

Este estimador es útil cuando se analizan variables categóricas, como la presencia o ausencia de una condición específica, como por ejemplo, altos niveles de estrés. Esto es relevante en nuestra pregunta de investigación para comparar grupos, o medir proporciones según condiciones como, altos niveles de estrés, o bajas horas de sueño.

Sesgo:
La proporción muestral es insesgada, ya que cada observación es una variable de Bernoulli con esperanza $p$:

$$E[\hat{p}] = p \implies \text{Sesgo}(\hat{p}) = E[\hat{p}] - p = 0$$

Consistencia:
Es consistente.

$\text{Var}(\hat{p}) = \frac{p(1-p)}{n}$, luego $\text{Var}(\hat{p}) \xrightarrow{n \to \infty} 0$. Con $p$ la proporción poblacional real.

Eficiencia:
Como es insesgada, su ECM se reduce a su varianza:

$$\text{ECM}(\hat{p}) = \text{Var}(\hat{p}) = \frac{p(1-p)}{n}$$

El coeficiente de correlación de Pearson consiste en cuantificar la relación lineal entre dos variables, y se define como:


$$\rho = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2} \, \sqrt{\sum_{i=1}^{n} (y_i - \bar{y})^2}}$$

Con:

- $\bar{x}$: La media muestral de cada dato $x_i$.
- $\bar{y}$: La media muestral de cada dato $y_i$.

Este estimador se alinea completamente con la pregunta de investigación planteada, ya que este nos permite medir relaciones lineales, lo que puede sugerir como impacta una variable por sobre otra.


Sesgo:
El coeficiente de correlación de Pearson es sesgado, con sesgo aproximado:

$$\text{Sesgo}(\hat{r}) \approx -\frac{\rho(1-\rho^2)}{2n}$$

Sin embargo, como nuestro dataset contiene $n = 400$, el sesgo es prácticamente despreciable.

Consistencia:
Es consistente. A medida que $n$ crece, $\hat{r}$ converge en probabilidad a $\rho$:

$$\hat{r} \xrightarrow{p} \rho \quad \text{cuando } n \to \infty$$

Eficiencia:
No existe una fórmula analítica simple para la varianza de $\hat{r}$, por lo que más adelante se justifica el uso del método Bootstrap para estimarla.

El coeficiente de correlación de Spearman consiste en cuantificar la fuerza y dirección de una relación monótona entre dos variables, y se define como:

$$\rho = \frac{6\sum_{i=1}^{n} d_i^2}{n \, (n^2 - 1)}$$

Con:

- ${d_i}$: La diferencia entre los dos rangos de cada observación.
- ${n}$: El número de observaciones

A diferencia del método de Pearson, este estimador permite medir relaciones no lineales de tipo monótonas, dado que clasifica por rangos a los valores de cada variable.  

Al igual que Pearson, es sesgado por la misma razón: es un cociente y la esperanza
no se distribuye directamente sobre una división. Sin embargo, se conoce que a medida que tenemos una muestra más grande, tiende a ser insesgado, es decir:

$$\lim_{n \to \infty} \text{Sesgo}(\hat{r}_s) = 0$$

Consistencia:
Es consistente. Converge en probabilidad a la correlación poblacional de Spearman $\rho_s$:

$$\hat{r}_s \xrightarrow{p} \rho_s \quad \text{cuando } n \to \infty$$

Eficiencia:
Al igual que Pearson, no existe una fórmula analítica simple para su varianza,
lo que nuevamente justifica el uso del método Bootstrap para estimarla más adelante.

```python
#Se importan las librerías a usar
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import beta as beta_dist

#Se carga el dataset al notebook
df = pd.read_csv("ScreenTime vs MentalWellness.csv")
df = df.drop("Unnamed: 15", axis=1)
df.head()
```

## Media de Horas en Pantalla
Dado que nuestra hipótesis vincula el consumo digital con el bienestar mental, estimar la media poblacional del tiempo en pantalla nos permite establecer el comportamiento promedio de la muestra y utilizarlo como variable de control para analizar el estrés.

Definición del Parámetro: Media poblacional del tiempo diario total de uso de pantalla (en horas) para la población de personas con hábitos digitales.

Estimador Puntual ($\bar{X}$): Calculado como el promedio aritmético de las horas en pantalla.

Insesgadez: Como la media muestral es un estimador insesgado, el valor obtenido representa la mejor aproximación sin desviaciones sistemáticas del parámetro poblacional.

Eficiencia: Al ser el estimador de mínima varianza para este tipo de datos, garantiza la máxima precisión posible con nuestros 400 datos.

```python
# Calculamos la media muestral (Estimador Puntual)
media_pantalla = df['screen_time_hours'].mean()

# Calculamos la desviación estándar muestral
std_pantalla = df['screen_time_hours'].std()

print(f"Estimador puntual de la media (x̄): {media_pantalla:.2f} horas/día")
print(f"Dispersión de la muestra (s): {std_pantalla:.2f} horas")

sns.histplot(df['screen_time_hours'], kde=True)
plt.title("Distribución de Horas en Pantalla")
plt.show()
```

## Proporción de niveles altos de estrés
El análisis exploratorio reveló que el estrés no se distribuye uniformemente, sino que se concentra en valores críticos ($\ge 8$). Estimar la proporción poblacional de este grupo es vital para entender la prevalencia de un estado de "alerta" o estrés severo en la población.
Calculado como la frecuencia ($\hat{p}$) relativa de individuos con stress_level_0_10 >= 8 sobre el total de la muestra ($n=400$).

Consistencia: Nuestro estimador es consistente, lo que significa que el error de estimación disminuye al aumentar el tamaño de nuestra muestra, garantizando que el valor se aproxime al real.

Insesgadez: Al igual que la media, la proporción muestral es un estimador insesgado de la proporción poblacional, ofreciendo una medida directa y sin sesgos de la realidad observada.

```python
# Definimos el umbral de 'alto estrés' según el análisis de la Entrega 1 (>= 8)
umbral_estres = 8

# Filtramos las filas que cumplen la condición
casos_altos = df[df['stress_level_0_10'] >= umbral_estres]

# Calculamos la proporción (casos de interés / total)
n = len(df)
proporcion_estres = len(casos_altos)/n

print(f"Total de personas con estrés >= 8: {len(casos_altos)}")
print(f"Estimador puntual de la proporción (p̂): {proporcion_estres:.2%}")

# Se grafican en un diagrama de pie los resultados
df['is_high_stress'] = df['stress_level_0_10'] >= umbral_estres
df['is_high_stress'].value_counts().plot(kind='pie', autopct='%1.1f%%', labels=['Alto Estrés', 'Estrés Moderado/Bajo'])
plt.title("Proporción de Estrés Alto (>=8)")
plt.show()
```

## Análisis Comparativo por Ocupación
Aquí realizamos un análisis multivariado. Agrupamos los datos por ocupación para detectar cómo el entorno (estudiante vs. empleado) influye en el uso de pantallas y el nivel de estrés.

```python
# Agrupar según ocupación
df['screen_time_hours'] = pd.to_numeric(df['screen_time_hours'], errors='coerce')
df['stress_level_0_10'] = pd.to_numeric(df['stress_level_0_10'], errors='coerce')

resumen_ocupacional = df.groupby('occupation').agg(
    media_pantalla=('screen_time_hours', 'mean'),
    proporcion_estres_alto=('stress_level_0_10', lambda x: (x >= 8).mean())
)

print("\n--- ANÁLISIS COMPARATIVO ---")
print(resumen_ocupacional.round(2))
```

Se puede observar, que en el grupo de los estudiantes, se presenta la mayor proporción de estrés alto y a su vez la mayor media de horas en pantalla.

## Diferencia entre proporciones de nivel de estrés según ocupación

Para comprender el impacto del estrés en los distintos grupos demográficos y visualizar cuáles son los segmentos más vulnerables, se evaluará la diferencia de proporciones de estrés crítico, es decir, mayor a 8, según cada ocupación. Se optó por utilizar proporciones en lugar de medias para evitar que la alta concentración de datos en los niveles máximos oculte el impacto real del problema.

El análisis se dividirá en dos segmentos poblacionales. El primero, la diferencia entre Empleados y Estudiantes, lo que permite comparar el nivel de tensión entre dos grupos sometidos a exigencias de rendimiento constante. El segundo, la diferencia entre Empleados y Desempleados, el cual contrastará la repercusión de enfrentar presiones y obligaciones diarias, frente al peso de la inactividad y la incertidumbre laboral.



El código a continuación permite dividir a la muestra en distintos dataframes separados por ocupación. Además se calcula la proporción del nivel de estrés crítico a través de la función proporción, que toma como parámetros un dataframe, el umbral de división, en este caso 8 y el nombre de la variable de la variable a estudiar, que corresponde al estrés.

```python
def proporcion(dataset, umbral, variable):
  contador = 0
  for valor in dataset[variable]:
      if(valor>=umbral):
          contador += 1
  return contador/len(dataset[variable])

# Agrupar ocupaciones
def agrupar_ocupacion(x):
    if x in ['Employed', 'Self-employed']:
        return 'Empleados'
    elif x in ['Unemployed', 'Retired']:
        return 'Desempleados'
    else:
        return 'Estudiantes'

df['grupo_ocupacion'] = df['occupation'].apply(agrupar_ocupacion)

# Separar en dataframes
df_empleados = df[df['grupo_ocupacion'] == 'Empleados']
df_desempleados = df[df['grupo_ocupacion'] == 'Desempleados']
df_estudiantes = df[df['grupo_ocupacion'] == 'Estudiantes']

# Calcular proporciones
proporcion_estres_emp = proporcion(df_empleados, 8, 'stress_level_0_10')
proporcion_estres_des = proporcion(df_desempleados, 8, 'stress_level_0_10')
proporcion_estres_est = proporcion(df_estudiantes, 8, 'stress_level_0_10')

print(f"Proporción de Estrés Crítico en Empleados: {proporcion_estres_emp:.3f}")
print(f"Proporción de Estrés Crítico en Desempleados: {proporcion_estres_des:.3f}")
print(f"Proporción de Estrés Crítico en Estudiantes: {proporcion_estres_est:.3f}")
```

A partir de la muestra obtenida, se calcularon los estimadores puntuales para

---

la proporción de estrés crítico en cada grupo demográfico, obteniendo un 0.627 (62.7%) para Empleados, un 0.748 (74.8%) para Estudiantes y un 0.268 (26.8%) para Desempleados.

# Bootstrap


## Explicación método de Bootstrap

Aún se mantiene el problema de que no podemos obtener información de la población completa, sin embargo, se posee una muestra representativa de esta. Si tenemos nuestro estimador puntual, estos tienen una incertidumbre asociada, debido a que no conocemos el valor exacto de la población.

Aquí, es donde entra el método de remuestreo de Bootstrap, donde en vez de recolectar información de la población completa, lo cual es muy difícil, Bootstrap simula nuevas muestras, a partir de la muestra previamente recolectada, usando remuestreo con reemplazo, es decir, podríamos seleccionar más de una vez el mismo dato para el remuestreo.

A continuación se detallan los pasos para realizar el método de remuestreo Bootstrap.

1.- Se toma la muestra con la que se quiere trabajar.

2.- De la muestra original de tamaño n, se realiza una remuestra de tamaño n, seleccionando aleatoriamente y con reemplazo, los índices de los datos con los que vamos a remuestrear.

3.- Se calcula el estimador de interés.

4.- Se repiten los pasos 2 y 3 las veces que se desee, el método Bootstrap usualmente se repite entre 1000 y 100000 veces.

5.- Por cada repetición del remuestreo, se guarda el valor del estimador requerido en cada iteración.

6.- Con los valores de remuestreos listos, se puede calcular o la media de los estimadores, su varianza o mapearlos en un histograma.

A continuación, se mostrarán las funciones que realizan el método de bootstrap, cabe destacar, que se decidió hacer funciones distintas para cada estimador que se desee remuestrear, pero todas mantienen la misma estructura de bootstrap, donde se separa la columna a remuestrear, se crea una lista vacía con el estimador remuestreado, se hace un bucle de 10000 iteraciones, donde se toma un arreglo de indices randomizados con valores de 0 a n con reemplazo, se extrae una lista con la remuestra según los índices obtenidos, se calcula el estimador deseado y se añade a la lista con los estimadores remuestreados, finalmente, se retorna dicha lista.

Se cargan las funciones

```python
def remuestreo_bootstrap_medias(dataset, n, columna):
    np.random.seed(777)
    datos = dataset[columna].values
    medias_remuestreo = []
    i=0
    while(i<10000):
        indices = np.random.randint(0,n, size=n)
        remuestras = datos[indices]
        medias_remuestreo.append(remuestras.mean())
        i=i+1
    return medias_remuestreo

def remuestreo_bootstrap_proporciones(dataset, n, columna, umbral):
    np.random.seed(777)
    datos = dataset[columna].values
    proporciones_remuestreo = []
    i=0
    while(i<10000):
        contador = 0
        indices = np.random.randint(0, n, size = n)
        remuestras = datos[indices]
        for valor in remuestras:
            if(valor>=umbral):
                contador = contador + 1
        proporciones_remuestreo.append(contador/n)
        i = i+1
    return proporciones_remuestreo

def remuestreo_bootstrap_diferencia_proporciones(dataset1, dataset2, columna, umbral):
    np.random.seed(777)
    datos1 = dataset1[columna].values
    datos2 = dataset2[columna].values
    dif_proporciones_remuestreo = []
    i=0
    while(i<10000):
        contador1 = 0
        contador2 = 0
        indices1 = np.random.randint(0, len(datos1), size = len(datos1))
        indices2 = np.random.randint(0, len(datos2), size = len(datos2))
        remuestras1 = datos1[indices1]
        remuestras2 = datos2[indices2]
        for valor in remuestras1:
            if(valor>=umbral):
                contador1 = contador1 + 1
        for valor in remuestras2:
            if(valor>=umbral):
                contador2 = contador2 + 1
        dif_proporciones_remuestreo.append(contador1/len(datos1) - contador2/len(datos2))
        i = i+1
    return dif_proporciones_remuestreo

def remuestreo_bootstrap_correlacion_pearson(dataset, n, columna1, columna2):
    np.random.seed(777)
    datos1 = dataset[columna1].values
    datos2 = dataset[columna2].values
    correlacion_pearson_remuestreo = []
    i=0
    while(i<10000):
        indices = np.random.randint(0, n, size=n)
        remuestras1 = datos1[indices]
        remuestras2 = datos2[indices]
        correlacion_pearson_remuestreo.append(np.corrcoef(remuestras1, remuestras2)[0, 1])
        i = i+1
    return correlacion_pearson_remuestreo

def remuestreo_bootstrap_correlacion_spearman(dataset, n, columna1, columna2):
    np.random.seed(777)
    datos1 = dataset[columna1].values
    datos2 = dataset[columna2].values
    correlacion_spearman_remuestreo = []
    i=0
    while(i<10000):
        indices = np.random.randint(0, n, size=n)
        remuestras1 = datos1[indices]
        remuestras2 = datos2[indices]
        correlacion_spearman_remuestreo.append(stats.spearmanr(remuestras1, remuestras2).statistic)
        i = i+1
    return correlacion_spearman_remuestreo
```

La utilidad del método Bootstrap, radica en que podemos conocer una distribución del estimador y una estimación de la incertidumbre de este, sin conocer o asumir la distribución de la variable original. Generalmente, si comparamos los resultados con estimadores puntuales con fórmula analítica como la media o la proporción, el valor estimado por Bootstrap es prácticamente idéntico al obtenido analíticamente, por lo que la diferencia no es significativa en términos en valor del estimador, sin embargo, para estimadores como los coeficientes de correlación, es difícil calcular su varianza con una fórmula, por lo que Bootstrap estima una varianza/incertidumbre de este tipo de estimadores.

Sin embargo, el método sigue conteniendo limitaciones, por ejemplo, depende totalmente de la muestra original, y si esta no es representativa de la población, Bootstrap simplemente reproduciría los errores que esta contenga.

# Simulación por Monte Carlo

## Explicación

El método Monte Carlo consiste en analizar fenómenos que involucran incertidumbre mediante la generación de un gran número de simulaciones aleatorias. La diferencia principal que tiene con bootstrap, es que el método de Monte Carlo, hace las remuestras con datos generados sintéticamente por modelos de distribución asumidos con parámetros estimados de la muestra personal (por ejemplo media o variaza).

Su propósito general es medir el comportamiento de algún sistema, gracias a la suposición de modelos, pudiendo ajustar sus parámetros. Esto se hila con su capacidad de explorar situaciones que no han sucedido en nuestras muestras reales, o, en vez de trabajar con un modelo asumido, buscar cuál es la mejor combinación de parámetros de las distribuciones para lograr un objetivo según lo plantee el investigador.

## Situaciones de interés

Con el método de simulación por Monte Carlo, es pertinente calcular las siguientes situaciones según nuestra pregunta de investigación.

- Se busca estimar la probabilidad, de que los estudiantes sean el grupo que padece de un mayor estrés crítico (>=8) en comparación con los empleados y los desempleados.
- ¿Cuál es la probabilidad de que el coeficiente de correlación de Pearson entre las horas en pantalla y los niveles de estrés sea mayor o igual al obtenido empíricamente si el fenómeno se repitiera más veces?

## Primera situación

### Modelo

Variables relevantes:
- Niveles de estrés.
- Proporción de niveles críticos de estrés.
- Grupos de ocupaciones.

Supuestos adoptados:
- Observando el gráfico, se usará la suposición de que los niveles de estrés siguen una distribución beta, esto es, debido a la asimetría que presentan los datos.
- Los niveles críticos de estrés, serán catalogados como aquellos que poseen niveles mayores o iguales que 8.

Método de simulación:
Se generarán datos aleatoriamente por cada ocupación, siguiendo la distribución beta, con alpha y beta calculados con fórmula analítica, la cual depende de la media y la varianza. Estos dos últimos valores, serán estimados por los obtenidos empíricamente. Se realizarán 10000 iteraciones y en cada iteración y se calculará la proporción de estrés crítico en cada ocupación.

- Observando el gráfico, se usará la suposición de que las horas en pantalla siguen una distribución normal, con medias y varianzas obtenidas empíricamente.

Método de simulación:
Se generarán datos aleatoriamente con la librería scipy y numpy. Siguiendo la distribución beta para los niveles de estrés con parámetros alpha y beta calculados con la media y varianza empírica. Las horas en pantalla, serán generadas aleatoriamente siguiendo una distribución normal, con los parámetros calculados análogamente a la distribución anterior.

Se define la función con la que se calcularán los coeficientes de correlación simulados.

```python
def monte_carlo_correlaciones(dataset, n_simulaciones):

    np.random.seed(60)

    # Parámetros para tiempo en pantalla
    media_pantalla = dataset['screen_time_hours'].mean()
    std_pantalla = dataset['screen_time_hours'].std()

    # Parámetros para nivel de estrés
    mu_estres = dataset['stress_level_0_10'].mean() / 10
    sigma2_estres = dataset['stress_level_0_10'].var() / 100
    t = (mu_estres * (1 - mu_estres) / sigma2_estres) - 1
    alpha_estres = mu_estres * t
    beta_estres = (1 - mu_estres) * t

    n = len(dataset)
    correlaciones_pantalla_estres = []

    for i in range(n_simulaciones):
        muestra_pantalla = np.random.normal(media_pantalla, std_pantalla, size=n)
        muestra_estres = beta_dist.rvs(alpha_estres, beta_estres, size=n) * 10

        corr_pantalla_estres = np.corrcoef(muestra_pantalla, muestra_estres)[0, 1]
        correlaciones_pantalla_estres.append(corr_pantalla_estres)

    return correlaciones_pantalla_estres
```

Para que el método de Monte Carlo fuese eficiente para este caso, debemos construirlo en base a una suposición, que sea asumiendo que los datos de distribuciones de horas en pantalla y niveles de estrés, sigan una distribución **conjunta**.

### Nuevas suposiciones y método

Se deben ajustar las suposiciones para aplicar Monte Carlo a esta situación.
- Dado que las horas en pantalla siguen una distribución normal, pero los niveles de estrés no, se inclinará la suposición hacia las horas en pantalla, es decir, se usará un modelo de distribución normal bivariada. Los parámetros de esta distribución, será la media y varianza correspondientes obtenidas empíricamente.

El nuevo método de simulación, será generar pares de números siguiendo la distribución bivariada mencionada anteriormente.

Se define la función, con la que se realizarán las remuestras siguiendo los supuestos recién comentados.

```python
def monte_carlo_normal_bivariada(df, columna1, columna2, n_simulaciones):

    np.random.seed(61)


    # Estimar parámetros desde los datos
    mu1 = df[columna1].mean()
    mu2 = df[columna2].mean()

    # Matriz de covarianza (se calcula, ya que es lo que recibe np.random.multivariate_normal)
    cov = np.cov(df[columna1], df[columna2])

    medias = [mu1, mu2]
    n = len(df)
    muestras = []

    for _ in range(n_simulaciones):
        muestra = np.random.multivariate_normal(medias, cov, size=n)
        muestras.append(muestra)

    return muestras
```

Se calculan las correlaciones y se agregan a una lista.

```python
remuestras=monte_carlo_normal_bivariada(df, 'stress_level_0_10', 'screen_time_hours', 10000)
r_original = np.corrcoef(df['screen_time_hours'], df['stress_level_0_10'])[0, 1]

correlaciones = []
for muestra in remuestras:
    corr = np.corrcoef(muestra[:, 0], muestra[:, 1])[0, 1]
    correlaciones.append(corr)

sns.histplot(correlaciones, bins=30, kde=True)
plt.xlabel('Coeficiente de correlación de Pearson')
plt.ylabel('Frecuencia')
plt.title('Distribución del coeficiente de correlación\nNormal Bivariada (Monte Carlo)')
plt.axvline(x=r_original, color='red', linestyle='--', label=f'Correlación original: {r_original:.4f}')
plt.axvline(x=np.mean(correlaciones), color='green', linestyle='--', label=f'Media simulada: {np.mean(correlaciones):.4f}')
plt.legend()
plt.show()
```

Por último, se calcula la probabilidad de que la correlación lineal entre los niveles de estrés y las horas en pantalla, superen el valor obtenido empíricamente.

```python
conteo = 0
for r in correlaciones:
    if r >= r_original:
        conteo += 1

probabilidad = conteo / len(correlaciones)

probabilidades_acumuladas = []

conteo = 0
for i, r in enumerate(correlaciones):
    if r >= r_original:
        conteo += 1
    probabilidades_acumuladas.append(conteo / (i + 1))

plt.plot(probabilidades_acumuladas)
plt.xlabel('Número de simulaciones')
plt.ylabel('Probabilidad acumulada')
plt.title(f'Estabilidad de P(r >= {r_original:.4f})')
plt.axhline(y=probabilidades_acumuladas[-1], color='red', linestyle='--', label=f'Valor final: {probabilidades_acumuladas[-1]:.4f}')
plt.legend()
plt.show()
```

Este resultado, es considerado alto, sobre todo por la condición de que lo iguale, o supere, es decir, los valores de correlacion obtenidos, tampoco se van a alejar tanto del valor obtenido empíricamente, por lo que si tenemos en cuenta estos supuestos, la correlación seguirá siendo positiva y aproximadamente mayor a 0.5, lo que es un valor de correlación alto para fenómenos sociales.
